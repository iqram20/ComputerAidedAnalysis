## 1. Lists and indexing

Lists are ordered and mutable. They are useful for small collections of engineering measurements.

```python
temperatures = [72.5, 74.1, 73.8, 75.0, 76.2]
print(temperatures[0])
print(temperatures[-1])
print(temperatures[1:4])
```

## 2. Tuples, sets, and dictionaries

- **Tuple:** ordered, typically used for fixed records such as coordinates.
- **Set:** keeps unique values.
- **Dictionary:** maps descriptive keys to values.

```python
sensor_xyz = (4.5, 3.2, 7.1)
ids = {101, 102, 101, 103}
motor = {"voltage_V": 240, "current_A": 12, "efficiency": 0.91}

print(sensor_xyz)
print(ids)
print(motor["efficiency"])
```

## 3. NumPy arrays

NumPy arrays support vectorized numerical operations. This is much more convenient than processing every measurement manually.

```python
import numpy as np

pressure_kPa = np.array([101.2, 102.5, 100.8, 103.1, 101.7])
pressure_Pa = pressure_kPa * 1000
print(pressure_Pa)
print("Mean pressure:", pressure_kPa.mean())
```

### 2D arrays

```python
forces = np.array([[10,20,30],
                   [40,50,60],
                   [70,80,90]])
print("Shape:", forces.shape)
print("Second row:", forces[1, :])
print("Third column:", forces[:, 2])
print("Diagonal:", np.diag(forces))
```

## 4. Engineering vectorization example

Calculate kinetic energy at several velocities at once.

```python
m = 5.0
v = np.array([2,4,6,8,10])
KE = 0.5 * m * v**2
print(KE)
```

## 5. Real open data: USGS earthquakes

```python
import pandas as pd
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_week.csv"
eq = pd.read_csv(url)
eq[["mag","depth","latitude","longitude","place"]].head()
```

Convert selected columns to NumPy arrays and inspect them.

```python
magnitudes = eq["mag"].dropna().to_numpy()
depths = eq["depth"].dropna().to_numpy()

print("First 10 magnitudes:", magnitudes[:10])
print("Magnitude array shape:", magnitudes.shape)
print("Mean magnitude:", np.mean(magnitudes))
```

### Worked Example A — Variables, types, and mathematical assignments

```python
import numpy as np

x = 10
y = 3

u = x + y
v = x * y
w = x / y
z = np.sin(x)
r = 8 * np.sin(x)
s = 5 * np.sin(x * y)
p = x**y

print(u, v, w, z, r, s, p)

S = "123"
N = float(S)
print(type(S), type(N))
```

### Worked Example B — Strings

```python
s1 = "HELLO"
s2 = "hello"

print(s1 == s2)
print(s1.lower() == s2)
print(s1 == s2.upper())

sentence = "Python is great!"
print("Python" in sentence)
print(sentence.split()[-1])
```

### Worked Example C — Lists, tuples, sets, dictionaries

```python
list_a = [1, 8, 9, 15]
list_a.insert(1, 2)
list_a.append(4)
list_a.sort()
print("List:", list_a)

tuple_a = ("One", 1)
print("Tuple second element:", tuple_a[1])

unique_values = set((2, 3, 2, 3, 1, 2, 5))
print("Unique:", unique_values)

set_a = {2, 3}
set_b = {1, 2, 3}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference:", set_a - set_b)

d = {"A":"a", "B":"b", "C":"c"}
print(d.keys())
print("B" in d)
```

### Worked Example D — NumPy arrays and Boolean indexing

```python
x = np.array([1, 4, 3, 2, 9, 4])
y = np.array([2, 3, 4, 1, 2, 3])

print("x+y =", x + y)
print("x*y =", x * y)
print("x/y =", x / y)
print("x**y =", x**y)

grid = np.linspace(-10, 10, 100)
print("Grid length:", len(grid))

array_a = np.array([-1, 0, 1, 2, 0, 3])
print("Positive values:", array_a[array_a > 0])
```

### Worked Example E — Matrix operations

```python
Y = np.array([
    [3, 5, 3],
    [2, 2, 5],
    [3, 8, 9]
])

print("Transpose:")
print(Y.T)

Z = np.zeros((2,4))
Z[:,1] = 1
print("Modified zero array:")
print(Z)
```
