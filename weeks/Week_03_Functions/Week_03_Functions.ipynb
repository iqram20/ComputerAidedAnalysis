## 1. Basic function structure

```python
def force(mass_kg, acceleration_m_s2):
    return mass_kg * acceleration_m_s2

F = force(20, 3.5)
print("Force =", F, "N")
```

## 2. Return is different from print

A returned result can be stored and reused in later calculations.

```python
def electrical_power(voltage, current):
    return voltage * current

P = electrical_power(24, 3.5)
energy_Wh = P * 2
print(P, energy_Wh)
```

## 3. Engineering function examples

```python
import numpy as np

def cylinder_volume(r, h):
    return np.pi * r**2 * h

def stress(force_N, area_m2):
    return force_N / area_m2

def kinetic_energy(m, v):
    return 0.5 * m * v**2

print(cylinder_volume(1.8, 5.5))
print(stress(12000, 0.004))
print(kinetic_energy(50, 10))
```

## 4. Lambda functions

A lambda is useful for a short one-expression function.

```python
circle_area = lambda r: np.pi * r**2
for r in [1, 2.5, 5]:
    print(r, circle_area(r))
```

## 5. Real open-data function lab: USGS earthquakes

```python
import pandas as pd
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_week.csv"
eq = pd.read_csv(url)
```

Create reusable functions instead of repeating analysis code.

```python
def dataset_summary(df):
    return {
        "count": len(df),
        "mean_mag": df["mag"].mean(),
        "max_mag": df["mag"].max(),
        "mean_depth_km": df["depth"].mean()
    }

def earthquakes_above(df, threshold):
    return df[df["mag"] >= threshold]

print(dataset_summary(eq))
strong = earthquakes_above(eq, 4.0)
strong[["time","mag","depth","place"]].head()
```

### Worked Example A — Hyperbolic sine function

```python
import numpy as np

def my_sinh(x):
    return (np.exp(x) - np.exp(-x)) / 2

for x in [0, 1, 2]:
    print(x, my_sinh(x), np.sinh(x))
```

### Worked Example B — Checkerboard array function

```python
def checkerboard(n):
    rows, cols = np.indices((n, n))
    return (rows + cols + 1) % 2

print(checkerboard(5))
```

### Worked Example C — Functions returning multiple results

```python
def cylinder_properties(r, h):
    surface_area = 2*np.pi*r**2 + 2*np.pi*r*h
    volume = np.pi*r**2*h
    return surface_area, volume

s, v = cylinder_properties(2, 4)
print("Surface area =", s)
print("Volume =", v)
```

### Worked Example D — Matrix splitting

```python
def split_matrix(m):
    ncols = m.shape[1]
    cut = (ncols + 1) // 2
    left = m[:, :cut]
    right = m[:, cut:]
    return left, right

M = np.arange(1, 10).reshape(3,3)
left, right = split_matrix(M)
print(left)
print(right)
```

### Worked Example E — Lambda and function-as-argument

```python
subtract = lambda x, y: x - y

def evaluate(func, a, b):
    return func(a, b)

print(evaluate(subtract, 10, 4))
```

### Worked Example F — Vectorized annulus area and tolerance search

```python
def annulus_area(r_inner, r_outer):
    r_inner = np.asarray(r_inner)
    r_outer = np.asarray(r_outer)
    return np.pi * (r_outer**2 - r_inner**2)

print(annulus_area(np.array([1,2,3]), np.array([2,4,6])))

def within_tolerance(A, target, tol):
    A = np.asarray(A)
    return np.where(np.abs(A - target) < tol)[0]

print(within_tolerance([0,1,2,3], 1.5, 0.75))
```
