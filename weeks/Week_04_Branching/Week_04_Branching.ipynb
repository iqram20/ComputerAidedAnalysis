## 5. Real open-data classification: USGS earthquakes

For this class exercise, use simple descriptive categories for programming practice.

```python
import pandas as pd
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_week.csv"
eq = pd.read_csv(url)

def classify_magnitude(m):
    if pd.isna(m):
        return "Unknown"
    elif m < 3:
        return "Small"
    elif m < 5:
        return "Moderate"
    elif m < 7:
        return "Strong"
    else:
        return "Major"

eq["magnitude_class"] = eq["mag"].apply(classify_magnitude)
eq[["mag","depth","place","magnitude_class"]].head(10)
```

### Worked Example A — Multi-level decision rule

```python
def service_level(load_percent):
    if load_percent < 60:
        return "Normal"
    elif load_percent < 80:
        return "Elevated"
    elif load_percent < 95:
        return "Warning"
    else:
        return "Critical"

for x in [45, 70, 87, 99]:
    print(x, service_level(x))
```

### Worked Example B — Operation selector

```python
def multi_operation(a, b, operation):
    if operation == "plus":
        return a + b
    elif operation == "minus":
        return a - b
    elif operation == "mult":
        return a * b
    elif operation == "div":
        return a / b
    elif operation == "pow":
        return a ** b
    else:
        raise ValueError("Unknown operation")

print(multi_operation(4, 2, "pow"))
```

### Worked Example C — Point classification inside a right triangle

```python
def point_in_unit_triangle(x, y):
    if x < 0 or y < 0 or x + y > 1:
        return "outside"
    elif x == 0 or y == 0 or x + y == 1:
        return "border"
    else:
        return "inside"

for p in [(0.5,0.5), (0.25,0.25), (5,5)]:
    print(p, point_in_unit_triangle(*p))
```

### Worked Example D — Redundant sensor alarm

```python
def sensor_alarm(s1, s2, s3, tolerance=10):
    disagreements = [
        abs(s1-s2) > tolerance,
        abs(s1-s3) > tolerance,
        abs(s2-s3) > tolerance
    ]

    if sum(disagreements) >= 1:
        return "alarm!"
    return "normal"

print(sensor_alarm(94, 96, 90))
print(sensor_alarm(94, 96, 80))
```

### Worked Example E — Quadratic-root cases

```python
import numpy as np

def quadratic_roots(a, b, c):
    D = b**2 - 4*a*c

    if D > 0:
        roots = [(-b + np.sqrt(D))/(2*a), (-b - np.sqrt(D))/(2*a)]
        return 2, roots
    elif D == 0:
        return 1, [-b/(2*a)]
    else:
        real = -b/(2*a)
        imag = np.sqrt(-D)/(2*a)
        return -2, [complex(real, imag), complex(real, -imag)]

print(quadratic_roots(1, 0, -9))
print(quadratic_roots(1, 2, 1))
print(quadratic_roots(1, 0, 9))
```
