## 1. Boolean comparisons

```python
temperature = 85
print(temperature > 80)
print(70 <= temperature <= 90)
print(temperature != 100)
```

## 2. Engineering alarm logic

```python
temperature = 85

if temperature < 70:
    status = "Normal"
elif temperature <= 90:
    status = "Warning"
else:
    status = "Critical"

print(status)
```

## 3. Combining conditions

```python
pressure_kPa = 480
temperature_C = 92

if pressure_kPa <= 500 and temperature_C <= 90:
    print("System within limits")
else:
    print("Inspect system")
```

## 4. Reusable classification function

```python
def classify_current(I):
    if I < 10:
        return "Normal"
    elif I <= 15:
        return "Warning"
    else:
        return "Trip circuit breaker"

for I in [6, 12, 18]:
    print(I, classify_current(I))
```

## 5. Real open-data classification: USGS earthquakes

For this class exercise, use simple descriptive categories for programming practice.

```python
import pandas as pd
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_week.csv"
eq = pd.read_csv(url)

def classify_magnitude(m):
    if pd.isna(m):
        return "Unknown"
    elif m < 3:
        return "Small"
    elif m < 5:
        return "Moderate"
    elif m < 7:
        return "Strong"
    else:
        return "Major"

eq["magnitude_class"] = eq["mag"].apply(classify_magnitude)
eq[["mag","depth","place","magnitude_class"]].head(10)
```
