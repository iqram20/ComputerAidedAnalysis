## 1. Python as an engineering calculator

Python follows normal mathematical precedence. Use parentheses to make engineering equations clear.

```python
2 + 3 * 4
(2 + 3) * 4
2**3
```

### Example: Ohm's law

For a resistor, **V = IR**. Store each physical quantity in a clearly named variable.

```python
current_A = 0.025
resistance_ohm = 220
voltage_V = current_A * resistance_ohm
print("Voltage =", voltage_V, "V")
```

### Example: kinetic energy

KE = 1/2 mv². Python uses `**2` for squaring.

```python
mass_kg = 1250
velocity_m_s = 22
KE_J = 0.5 * mass_kg * velocity_m_s**2
print(f"Kinetic energy = {KE_J:,.1f} J")
```

## 2. NumPy mathematical functions

NumPy provides constants and scientific functions used throughout engineering.

```python
import numpy as np

radius_m = 1.8
height_m = 5.5
volume_m3 = np.pi * radius_m**2 * height_m
print(f"Cylinder volume = {volume_m3:.2f} m^3")

angle_deg = 30
angle_rad = np.deg2rad(angle_deg)
print("sin(30°) =", np.sin(angle_rad))
```

## 4. Real open data: USGS earthquakes

USGS publishes a continuously updated public CSV feed. Here we use the **past 7 days** feed. The columns include earthquake magnitude, coordinates, depth, and place.

```python
import pandas as pd

url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_week.csv"
eq = pd.read_csv(url)
print("Rows:", len(eq))
eq.head()
```

At this stage, focus only on simple arithmetic with columns. pandas will be covered more deeply later.

```python
avg_mag = eq["mag"].mean()
max_mag = eq["mag"].max()
avg_depth = eq["depth"].mean()

print(f"Average magnitude: {avg_mag:.2f}")
print(f"Maximum magnitude: {max_mag:.2f}")
print(f"Average depth: {avg_depth:.2f} km")
```
