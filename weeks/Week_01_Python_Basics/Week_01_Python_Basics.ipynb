### Example: Ohm's law

For a resistor, **V = IR**. Store each physical quantity in a clearly named variable.

```python
current_A = 0.025
resistance_ohm = 220
voltage_V = current_A * resistance_ohm
print("Voltage =", voltage_V, "V")
```

### Example: kinetic energy

KE = 1/2 mv². Python uses `**2` for squaring.

```python
mass_kg = 1250
velocity_m_s = 22
KE_J = 0.5 * mass_kg * velocity_m_s**2
print(f"Kinetic energy = {KE_J:,.1f} J")
```

### Worked Example A — Geometry calculations

Use Python to calculate:
- triangle area,
- cylinder surface area,
- cylinder volume,
- slope between two points,
- distance between two points.

```python
import math

# Triangle
base = 10
height = 12
triangle_area = 0.5 * base * height
print("Triangle area =", triangle_area)

# Cylinder
r = 5
h = 3
surface_area = 2 * math.pi * r**2 + 2 * math.pi * r * h
volume = math.pi * r**2 * h
print("Cylinder surface area =", surface_area)
print("Cylinder volume =", volume)

# Two points
x1, y1 = 3, 4
x2, y2 = 5, 9

slope = (y2 - y1) / (x2 - x1)
distance = math.sqrt((x2-x1)**2 + (y2-y1)**2)

print("Slope =", slope)
print("Distance =", distance)
```

### Worked Example B — Scientific functions

```python
import math

print("6! =", math.factorial(6))

x = 2
sinh_from_definition = (math.exp(x) - math.exp(-x)) / 2
print("sinh from definition =", sinh_from_definition)
print("math.sinh =", math.sinh(x))

angles = [math.pi, math.pi/2, math.pi/4, math.pi/6]
for a in angles:
    value = math.sin(a)**2 + math.cos(a)**2
    print(a, value)

print("sin(87 degrees) =", math.sin(math.radians(87)))
```

### Worked Example C — Leap-year logic

A leap year is divisible by 4, except century years must also be divisible by 400.

```python
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

count = 0
for year in range(1500, 2011):
    if is_leap_year(year):
        count += 1

print("Leap years from 1500 through 2010 =", count)
```

### Worked Example D — Ramanujan-style π approximation

This example practices factorials, powers, summation, and comparison with Python's stored π.

```python
import math

def ramanujan_pi(N):
    total = 0
    for k in range(N + 1):
        numerator = math.factorial(4*k) * (1103 + 26390*k)
        denominator = math.factorial(k)**4 * 396**(4*k)
        total += numerator / denominator
    inv_pi = (2 * math.sqrt(2) / 9801) * total
    return 1 / inv_pi

for N in [0, 1]:
    approx = ramanujan_pi(N)
    print("N =", N, "approx =", approx, "error =", abs(approx - math.pi))
```

### Worked Example E — Logical expressions and truth tables

```python
values = [True, False]

print("Law of noncontradiction")
for P in values:
    print(P, P and (not P))

print("\nDe Morgan check")
for P in values:
    for Q in values:
        left1 = not (P or Q)
        right1 = (not P) and (not Q)
        left2 = not (P and Q)
        right2 = (not P) or (not Q)
        print(P, Q, left1 == right1, left2 == right2)
```
