## 1. for loops

```python
for t in range(0, 11):
    a = 3.0
    v = a * t
    print(f"t={t:2d} s, v={v:5.1f} m/s")
```

## 2. while loops: tank filling

```python
volume_L = 0
flow_L_min = 12
minutes = 0

while volume_L < 500:
    volume_L += flow_L_min
    minutes += 1

print(minutes, volume_L)
```

## 3. Cooling simulation

```python
T = 100
minute = 0
while T >= 40:
    print(minute, T)
    T -= 5
    minute += 1
print("Below 40 C after", minute, "minutes")
```

## 4. List comprehensions

```python
velocities = [2,4,6,8,10]
KE = [0.5 * 5 * v**2 for v in velocities]
print(KE)
```

## 5. Real open-data loop lab: USGS earthquakes

```python
import pandas as pd
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_week.csv"
eq = pd.read_csv(url)
```

Although pandas can count quickly, loops help us understand algorithmic thinking.

```python
count_ge4 = 0
max_mag = float("-inf")
max_place = None

for _, row in eq.iterrows():
    m = row["mag"]
    if pd.isna(m):
        continue
    if m >= 4:
        count_ge4 += 1
    if m > max_mag:
        max_mag = m
        max_place = row["place"]

print("M >= 4:", count_ge4)
print("Largest:", max_mag, max_place)
```

## 6. Mini engineering simulation

Simulate a battery that starts at 100% state of charge and loses 7% each hour until it falls below 20%. Print a warning when it first falls below 30%.

```python
soc = 100
hour = 0
while soc >= 20:
    print(hour, soc)
    if soc < 30:
        print("Warning: recharge soon")
    soc -= 7
    hour += 1
```
